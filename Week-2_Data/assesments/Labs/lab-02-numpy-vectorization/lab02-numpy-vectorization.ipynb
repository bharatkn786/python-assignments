## Task 1: Creating a 2D NumPy Array

### Step 1.1: Import NumPy

Bring in NumPy so array operations are available.



```python
import numpy as np
```

### Step 1.2: Build a 2D Array

Set up a 2D array to work with for the rest of the exercises.

```python
data = np.array([
    [10, 20, 30],
    [15, 25, 35],
    [20, 30, 40],
    [25, 35, 45]
])
data
```

### Step 1.3: Compute Row Means and Column Maxima

Use NumPy's built-in aggregation functions to get the average of each row and the largest value in each column.


row_means = np.mean(data, axis=1)
column_maxima = np.max(data, axis=0)

row_means, column_maxima


## Task 2: Normalize Each Column

### Step 2.1: Scale Columns to 0–1

Rescale every column so its values fall between 0 and 1, all without writing an explicit loop.


```python
column_min = data.min(axis=0)
column_max = data.max(axis=0)

normalized_data = (data - column_min) / (column_max - column_min)

normalized_data
```

## Task 3: Applying Broadcasting

### Step 3.1: Subtract Column Means

Use NumPy broadcasting to subtract column means from every row.


```python
column_means = data.mean(axis=0)

centered_data = data - column_means

centered_data
```

## Task 4: Compare Vectorized Operation with Python Loop

### Step 4.1: Compare Vectorized Operation with Python Loop

Compare the execution time of a NumPy vectorized operation with an equivalent Python loop



```python
import time

large_data = np.random.rand(1000000)
```

```python
start = time.time()

vectorized_result = large_data * 2

vectorized_time = time.time() - start

print("Vectorized Time:", vectorized_time)
```

```python
start = time.time()

loop_result = []

for value in large_data:
    loop_result.append(value * 2)

loop_time = time.time() - start

print("Loop Time:", loop_time)
```

```python
print("Speed Difference:", loop_time / vectorized_time)
```


## Overall Observations

- NumPy handles numerical computation through efficient array operations rather than element-by-element processing.
- Vectorized code runs noticeably faster than the equivalent Python `for` loop — in this case, over 22x faster.
- Broadcasting lets arrays of different shapes interact directly, removing the need for manual looping.
- Normalizing columns down to a 0–1 range puts every feature on a comparable scale, which is useful before further analysis.
- Built-in aggregation functions like `mean()` and `max()` make common statistical calculations a one-line affair.
