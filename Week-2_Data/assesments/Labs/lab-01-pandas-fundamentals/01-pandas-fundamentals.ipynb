
## Task 2: Selecting Rows and Columns

***Step 2.1: Select Columns***

Pull out a single column or a subset of columns for targeted analysis.



````python
df["alcohol"]
````

````python
df[["alcohol", "quality"]]
````

***Step 2.2: Select Rows with `loc[]`***

`loc[]` selects rows and columns by their labels rather than their positions.

````python
df.loc[0]
````

````python
df.loc[0:5, ["alcohol", "quality"]]
````

***Step 2.3: Select Rows with `iloc[]`***

`iloc[]` does the same job as `loc[]`, but works off integer positions instead of labels.



````python
df.iloc[0]
````

````python
df.iloc[0:6, [11, 12]]
````

***Step 2.4: Filter Rows with Boolean Conditions***

Apply a condition to a column and use it as a mask to keep only the matching rows.

````python
df[df["quality"] > 7]
````

````python
df[df["alcohol"] > 12]
````


### Task 3: Create a New Column

***Step 3.1: Derive a Boolean Column***

Build a new column that flags whether alcohol content crosses a threshold.


````python
df["alcohol_level"] = df["alcohol"] > 15
````

````python
df[["alcohol", "alcohol_level"]].head()
````

***Step 3.2: Derive a Numeric Column***

Compute a ratio column by dividing one numeric column by another.

````python
df["acidity_ratio"] = df["fixed acidity"] / df["volatile acidity"]
````

````python
df[["fixed acidity", "volatile acidity", "acidity_ratio"]].head()
````

### Why No Loop?

You might think we need:

***for row in df:***
    ...

    But we don't.

Pandas performs the operation across the entire column:

***df["fixed acidity"] / df["volatile acidity"]***
    
This is vectorized computation.

This concept becomes especially important in my NumPy lab.

### Task 4: Group By and Aggregate

***Step 4.1: Group by Quality***

Group rows by the `quality` category to prepare for per-group calculations.


````python
df.groupby("quality")
````

***Step 4.2: Aggregate Mean and Count***

For each quality group, compute the average alcohol content and how many rows fall into it.



````python
df.groupby("quality")["alcohol"].agg(["mean", "count"])
````

### Task 5: Merge DataFrames

***Step 5.1: Build Two Sample DataFrames***

Set up two small DataFrames that share an `Id` column, so they can be joined together.


````python
df1 = pd.DataFrame({
    "Id": [1, 2, 3, 4],
    "quality": [5, 6, 7, 8]
})

df2 = pd.DataFrame({
    "Id": [1, 2, 3, 4],
    "category": ["Low", "Medium", "Good", "Excellent"]
})
````

***Step 5.2: Merge on the Shared Key***

Join the two DataFrames on `Id` to combine their columns into one table

````python
merged_df = pd.merge(df1, df2, on="Id")
merged_df
````

***Step 5.3: Check the Row Count***

Confirm the merge produced the expected number of rows.



````python
merged_df.shape[0]
````


## Overall Observations

- Pandas makes it straightforward to load, inspect, and manipulate tabular data with just a few method calls.
- Functions like `head()`, `info()`, `describe()`, and `shape` give a fast first impression of a dataset's size, types, and spread.
- Rows and columns can be accessed through plain indexing, `loc[]` for label-based selection, or `iloc[]` for position-based selection.
- Boolean masks make it easy to isolate rows that meet a specific condition.
- New columns can be built on the fly from existing ones, whether as flags or calculated ratios.
- `groupby()` combined with `agg()` is a compact way to summarize data across categories.
- `merge()` lets separate DataFrames be stitched together using a common key column.
