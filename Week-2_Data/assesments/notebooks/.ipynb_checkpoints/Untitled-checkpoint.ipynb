# Pandas and NumPy

Pandas and NumPy are two important Python libraries used for working with data.

NumPy is mainly focused on numerical computing and provides efficient array structures for calculations.

Pandas builds on these ideas by providing labeled data structures such as Series and DataFrames, which make it easier to work with tabular datasets.

This notebook covers the basic operations needed for numerical processing and data manipulation.

### Learning Objectives

- Understand NumPy arrays and their properties
- Perform numerical operations using NumPy
- Understand Pandas Series and DataFrames
- Load and inspect datasets using Pandas
- Select, filter, and modify data
- Perform basic data processing operations

```python
import numpy as np
import pandas as pd
```

## What is NumPy?

NumPy stands for Numerical Python. It is designed to handle numerical data efficiently using multidimensional arrays.

A NumPy array stores values in a structured format and allows operations to be performed on many elements at the same time.

Some useful features of NumPy include:

- Fast numerical calculations
- One-dimensional and multidimensional arrays
- Mathematical and statistical functions
- Array indexing and slicing
- Reshaping and filtering

```python
scores = np.array([12, 24, 36, 48, 60])

print("Array:", scores)
print("Type:", type(scores))
print("Shape:", scores.shape)
print("Dimensions:", scores.ndim)
print("Data type:", scores.dtype)
```

### NumPy Array Operations

```python
arr = np.array([10, 20, 30, 40, 50])

print("Addition:", arr + 5)
print("Subtraction:", arr - 5)
print("Multiplication:", arr * 2)
print("Division:", arr / 10)
```

### NumPy Indexing and Slicing

Indexing is used to access individual elements of an array, while slicing is used to access a range of elements.

```python
print("First element:", arr[0])
print("Last element:", arr[-1])
print("First three elements:", arr[:3])
print("Elements from index 2:", arr[2:])
```

### NumPy 2D Arrays

A 2D NumPy array consists of rows and columns and can be used to represent tabular or matrix-like data.

The structure of a 2D array can be examined using properties such as shape, ndim, and size.

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(data)
print("Shape:", data.shape)
print("Dimensions:", data.ndim)
print("Size:", data.size)
```

### Indexing and Slicing in 2D Arrays

In a 2D array, an element can be accessed using its row and column index. Python uses zero-based indexing

```python
print("First element:", data[0, 0])
print("Second row:", data[1])
print("Second column:", data[:, 1])
print("First two rows:")
print(data[:2])
```

### Statistical Operations with NumPy

NumPy provides built-in functions for common statistical calculations such as mean, median, minimum, maximum, and standard deviation.

```python
values = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(values))
print("Median:", np.median(values))
print("Minimum:", np.min(values))
print("Maximum:", np.max(values))
print("Standard deviation:", np.std(values))
```

### Reshaping NumPy Arrays

Reshaping changes the structure of an array without changing its data.

The total number of elements must remain the same before and after reshaping

```python
arr = np.arange(1, 13)

print("Original array:")
print(arr)

reshaped = arr.reshape(3, 4)

print("\nReshaped array:")
print(reshaped)

print("\nNew shape:", reshaped.shape)
```

### Conditional Filtering

NumPy allows us to filter array elements using conditions.

This is useful when we need to select observations that satisfy a specific condition.

```python
values = np.array([10, 25, 40, 15, 60, 30])

filtered = values[values > 30]

print("Original values:", values)
print("Values greater than 30:", filtered)
```

### NumPy Summary

In this section, we learned:

- NumPy arrays and `ndarray`
- Array properties such as shape, dimensions, size, and data type
- Array indexing and slicing
- 2D arrays
- Mathematical and statistical operations
- Array reshaping
- Conditional filtering

These operations form the foundation for numerical data processing in Python.

## What is Pandas?

Pandas is a Python library used for data manipulation and analysis.

It provides two important data structures:

- **Series:** One-dimensional labeled data
- **DataFrame:** Two-dimensional tabular data

```python
import pandas as pd

series = pd.Series([10, 20, 30, 40, 50])

print(series)
print("\nType:", type(series))
```

### Pandas Series Indexing

A Series can be accessed using its index.

By default, Pandas assigns integer indexes starting from 0

```python
print("First value:", series[0])
print("Second value:", series[1])
print("Last value:", series[4])
```

```python
marks = pd.Series(
    [85, 90, 78, 92],
    index=["Maths", "Physics", "Chemistry", "English"]
)

print(marks)
print("\nPhysics marks:", marks["Physics"])
```

### What is a DataFrame?
A DataFrame is a two-dimensional labeled data structure in Pandas.

It is similar to a table with rows and columns.

DataFrames are commonly used for storing, manipulating, and analyzing datasets.

```python
data = {
    "Name": ["Aman", "Riya", "Kartik", "Neha"],
    "Age": [21, 22, 20, 23],
    "Marks": [85, 92, 78, 88]
}

df_demo = pd.DataFrame(data)

print(df_demo)
```

## Inspecting a DataFrame

Pandas provides useful functions to quickly understand a dataset.

- `head()` shows the first rows.
- `tail()` shows the last rows.
- `info()` shows column names, data types, and non-null values.
- `describe()` gives a statistical summary of numerical columns.

```python
df = pd.read_csv("../../data/WineQT (3).csv")
print(df.head())
```

```python
print(df.tail())
```

```python
df.info()
```

```python
df.describe()
```

### Selecting Columns
Pandas allows us to select one or more columns from a DataFrame.

Selecting specific columns is useful when we want to focus on particular features during data analysis.

```python
# Select one column
print(df["alcohol"])
```

```python
# Select multiple columns
selected = df[["alcohol", "pH", "quality"]]

print(selected.head())
```

### Selecting Rows
Rows can be selected using conditions.

This is useful for filtering observations based on specific requirements.

```python
# Wines with quality greater than 6
high_quality = df[df["quality"] > 6]

print(high_quality.head())
```

### Selecting Rows and Columns with loc and iloc
loc is used for label-based selection.

iloc is used for position-based selection using integer indexes.

```python
print(df.loc[0, "alcohol"])
print(df.loc[0:4, ["alcohol", "pH", "quality"]])
```

```python
# Using iloc
print(df.iloc[0, 10])
print(df.iloc[0:5, 10:13])
```

### Filtering with Multiple Conditions

Pandas allows us to filter rows using multiple conditions.

- `&` means **AND**
- `|` means **OR**
- Each condition should be written inside parentheses.

```python
# AND condition
filtered = df[(df["quality"] >= 6) & (df["alcohol"] > 10)]

print(filtered.head())
```

```python
# OR condition
filtered = df[(df["quality"] >= 7) | (df["alcohol"] > 13)]

print(filtered.head())
```

```python
# Multiple columns with conditions
filtered = df[
    (df["quality"] >= 6) &
    (df["pH"] < 3.5) &
    (df["alcohol"] > 10)
]

print(filtered.head())
```

Sorting DataFrame
Pandas allows us to sort rows based on one or more columns.

sort_values() is used to arrange data in ascending or descending order.

```python
# Sort by alcohol in ascending order
sorted_df = df.sort_values("alcohol")

print(sorted_df[["alcohol", "quality"]].head())
```

```python
# Sort by alcohol in descending order
sorted_df = df.sort_values("alcohol", ascending=False)

print(sorted_df[["alcohol", "quality"]].head())
```

### Creating a New Column
A new column can be created by assigning a value or calculation to a new column name.

```python
df["alcohol_percentage"] = df["alcohol"]

print(df[["alcohol", "alcohol_percentage"]].head())
```
